```python
openai.api_key = "REDACTED"

def chit_chat_gold_summary(row):
    chat = row.get("chat")
    persona = row.get("Persona")
    if pd.isna(chat):
        return "none"
    if pd.notna(chat):
        # Prompt for rows with image captions
        prompt = (
            "Summarize the following conversation history concisely while preserving and keeping in mind key details such as persona of the user. Output only the summary in a paragraph.\n "
            f"Persona: {persona}\n"
            f"Conversation: {chat}\n"
            "Summary: \n"
        )
        #print(prompt)

    try:
        response = openai.ChatCompletion.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": "You are an expert in summarization of chit chat conversations"},
                {"role": "user", "content": prompt}
            ],
            max_tokens=300,
            temperature=0.3,
        )
        summary = response.choices[0].message.content.strip()
        return summary
    except Exception as e:
        print(f"Error during API call: {e}")
        return "Error"

def process_dataset(csv_file_path):
    df = pd.read_csv(csv_file_path)
    required_columns = ["Persona", "chat"]
    for col in required_columns:
        if col not in df.columns:
            df[col] = None

    df = df.head(2000)
    # print(df)
    results = []
    i=0
    for _, row in tqdm(df.iterrows(), desc="Generating Summaries", total=len(df)):
        summary = chit_chat_gold_summary(row)
        i+=1
        chat = row.get("chat")
        persona = row.get("Persona")
        results.append({
            "conversation number": i,
            "persona": persona,
            "chat": chat,
            "summary": summary
        })
        print(f"Conversation Number: {i}\nSummary: {summary}\n{'-'*50}")

    results_df = pd.DataFrame(results)
    output_file_path = csv_file_path.replace(".csv", "_gold_summary.csv")
    results_df.to_csv(output_file_path, index=False)
    print(f"Generated summaries have been extracted and saved to {output_file_path}")

csv_file_path = "/content/personality.csv"
process_dataset(csv_file_path)
```
